```python
import os
import numpy as np
from pprint import pprint


## python .\examples\new-pattern\new-pattern.py --P 3 --N 9 --num_new_neurons 6 --M 2000 --time 2000 --inference_time 1 --display True --result_dir results --image_name 'new-pattern-hist-P3N9-P2N15(2).png'
```

```python
results = "results"
import matplotlib.pyplot as plt
import json

png_files = [f for f in os.listdir(results) if f.endswith('.png')]
for file in png_files:
    json_name = file[:-4] + '.json'
    img = plt.imread(os.path.join(results, file))
    plt.figure(figsize=(10, 6))  # Make the figure size larger
    plt.imshow(img)
    plt.show()
    try:
        with open(os.path.join(results, json_name), 'r') as f:
            data = json.load(f)
            data = {k: set(v) for k,v in data.items()}
            pprint(data)
    except:
        pass

```
